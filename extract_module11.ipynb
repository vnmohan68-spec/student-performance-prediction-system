```python
# Extract Module 11.docx Code
This notebook loads `Module 11.docx` using `python-docx` and extracts the text and structure for review.
</VSCode.Cell>
<VSCode.Cell id="c2" language="python">
# Import Required Libraries
from pathlib import Path
import pandas as pd
try:
    import docx
except ImportError:
    import sys
    import subprocess
    subprocess.check_call([sys.executable, '-m', 'pip', 'install', 'python-docx'])
    import docx

print('python-docx version:', docx.__version__)
</VSCode.Cell>
<VSCode.Cell id="c3" language="python">
# Load Word Document
docx_path = Path('Module 11.docx')
assert docx_path.exists(), f'{docx_path} not found'
document = docx.Document(docx_path)
print('Loaded document:', docx_path.name)
print('Paragraph count:', len(document.paragraphs))
print('Table count:', len(document.tables))
</VSCode.Cell>
<VSCode.Cell id="c4" language="python">
# Extract Paragraph Text
paragraph_text = [p.text.strip() for p in document.paragraphs if p.text.strip()]
paragraph_df = pd.DataFrame({'paragraph_index': list(range(1, len(paragraph_text) + 1)), 'text': paragraph_text})
paragraph_df.head(20)
</VSCode.Cell>
<VSCode.Cell id="c5" language="python">
# Analyze Document Structure
structure = {
    'num_paragraphs': len(document.paragraphs),
    'num_nonempty_paragraphs': len(paragraph_text),
    'num_tables': len(document.tables),
    'paragraph_lengths': [len(t) for t in paragraph_text],
}
print(structure)
</VSCode.Cell>
<VSCode.Cell id="c6" language="python">
# Save Extracted Data
output_text_path = Path('module11_paragraphs.txt')
output_csv_path = Path('module11_paragraphs.csv')
paragraph_df.to_csv(output_csv_path, index=False)
output_text_path.write_text('\n\n'.join(paragraph_text), encoding='utf-8')
print('Saved extracted paragraphs to', output_csv_path)
print('Saved plain text output to', output_text_path)
```
